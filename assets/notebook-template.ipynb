```python
from pathlib import Path
import sys

# 从当前目录向上寻找这套课程的清单。
candidates = [Path.cwd(), *Path.cwd().parents]
ROOT = next((folder for folder in candidates if (folder / 'course.json').is_file()), None)
if ROOT is None:
    raise FileNotFoundError('请从课程文件夹或 notebooks 文件夹启动 Jupyter，并确认存在 course.json。')
print('课程位置：', ROOT)
print('当前 Python：', sys.executable)
```

## 解释准备单元

`Path.cwd()` 取得当前工作目录，`parents` 列出上级目录。我们通过 `course.json` 识别课程根目录；`ROOT` 只是当前 Notebook 的变量。请核对打印的 Python 是否属于本课环境。
